**My Lane:** Lane 2 - User Engagement & CTR Prediction

**Why I chose this:**
I want to understand what makes people click on search results. This is a fundamental question that directly relates to the data we have. ## 2. The question: decision, action, cost of a wrong call

**The Four Questions (from the framing-ml-problems skill):**

**1. What decision does this improve?**
Which pages should a content creator prioritize improving to get more clicks? Instead of guessing which content to fix, they can use a prediction to decide which pages need attention first.

**2. Who acts on the output, and what do they do?**
A content creator or website owner. They see the predicted CTR and decide:
- If CTR is low → improve the page
- If CTR is high → leave it alone or use it as a template

**3. What does a wrong answer cost?**
- **False Positive** (predict high CTR, actually low): Creator wastes time fixing content that won't perform well
- **False Negative** (predict low CTR, actually high): Creator misses an opportunity to improve something that already works

The cost is moderate - wasted time and missed opportunities, not life-or-death decisions.

**4. Why does data or ML help at all?**
Position alone already has a strong relationship with CTR, but other factors like content type, content age, and impressions also matter. The pattern is real but too messy to write as a simple rule. ML can find the combination of signals that best predicts CTR, and can adapt as patterns change over time.

**My Research Question:**
What factors most strongly predict whether a user will click on a search result?

## 3. Quick look at the data (2-3 real numbers)

```python
# Load the data and show supporting numbers
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the starter dataset
df = pd.read_csv('data/sample_data.csv')
print("Data loaded successfully!")
print(f"Data shape: {df.shape}\n")

print("="*60)
print("SUPPORTING NUMBERS FOR MY LANE")
print("="*60)

# Number 1: Position vs CTR
print("\n1. Position vs CTR:")
position_ctr = df.groupby('avg_position')['ctr'].mean()
print(position_ctr.head(10))

pos1_ctr = df[df['avg_position'] == 1]['ctr'].mean()
pos2_ctr = df[df['avg_position'] == 2]['ctr'].mean()
print(f"\nPosition 1 CTR: {pos1_ctr:.3f}")
print(f"Position 2 CTR: {pos2_ctr:.3f}")
print(f"Drop from position 1 to 2: {(pos1_ctr - pos2_ctr):.3f} ({(pos1_ctr - pos2_ctr) / pos1_ctr * 100:.1f}% drop)")

# Number 2: Content Type Performance
print("\n" + "="*60)
print("2. Content Type Performance")
print("="*60)

if 'content_type' in df.columns:
    content_ctr = df.groupby('content_type')['ctr'].mean().sort_values(ascending=False)
    print(content_ctr)
    print(f"\nBest: {content_ctr.index[0]} with CTR {content_ctr.iloc[0]:.3f}")
    print(f"Worst: {content_ctr.index[-1]} with CTR {content_ctr.iloc[-1]:.3f}")

# Number 3: Content Age Impact
print("\n" + "="*60)
print("3. Content Age Impact")
print("="*60)

if 'content_age_days' in df.columns:
    fresh = df[df['content_age_days'] <= 7]['ctr'].mean()
    old = df[df['content_age_days'] > 30]['ctr'].mean()
    print(f"Fresh content (≤ 7 days): {fresh:.3f}")
    print(f"Old content (> 30 days): {old:.3f}")
    print(f"Fresh vs old: {fresh - old:.3f} difference")

print("\n" + "="*60)
print("WHAT THESE NUMBERS TELL ME")
print("="*60)

print(f"""
1. Position matters: Position 1 has {pos1_ctr:.3f} CTR vs Position 2 has {pos2_ctr:.3f} CTR.
   This is a clear pattern worth predicting.

2. Content type matters: Different content types perform differently.
   This means creators can take action based on what works.

3. Freshness matters: Newer content tends to get more clicks.
   This gives a clear action: keep content fresh!

These patterns show this lane is worth exploring further because there are
real, measurable differences that ML could help predict.
""")
```
